1. . Largest Difference in List
Problem:

Given a list of integers, return the largest difference between any two elements, where the
larger element appears after the smaller one. If no such pair exists, return 0.

Example:
Input: [1, 2, 90, 10, 110]
Output: 109

Explanation: The largest difference is between 1 and 110.
make  python code
get the diff between two number
list is [1, 2, 90, 10, 110]

```python
def largest_difference(nums):
    if not nums or len(nums) < 2:
        return 0

    min_element = nums[0]
    max_diff = 0

    for num in nums[1:]:
        if num > min_element:
            max_diff = max(max_diff, num - min_element)
        else:
            min_element = num

    return max_diff

print(largest_difference([1, 2, 90, 10, 110]))
print(largest_difference([5, 4, 3, 2, 1]))
print(largest_difference([3, 1, 4, 8, 2, 10]))
```

2. First Non-Repeating Character
Problem:

Given a string, return the first non-repeating character. If there are no non-repeating characters,
return None.

Example:
Input: "aabcc"
Output: 'b'
Explanation: The first non-repeating character is 'b'
.

```python
def first_non_repeating_char(str_text):
    char_count = {}
    for char in str_text:
        char_count[char] = char_count.get(char, 0) + 1

    for char in str_text:
        if char_count[char] == 1:
            return f'First Non-Repeated Character is: {char}'

    return 'No Non-Repeated Character Found'

# Test cases
print(first_non_repeating_char("aabcc"))
print(first_non_repeating_char("aabccffggkkllb"))
```

3. Check If Strings Are Permutations

Problem:
Given two strings, write a function to check if one is a permutation of the other. Return True if
they are permutations, False otherwise.

Example:

Input: "abc", "bca"

Output: True

Explanation: Both strings contain the same characters in any order.

Note:-
Permutation is an arrangement of objects in a specific order

```python
def are_permutation(str1, str2):
  # If lengths are not equal, they can't be permutations
  if len(str1) != len(str2):
    print(f' {str1} and {str2} are NOT permutations of each other')
    return False

  else:

    # Sort both strings and compare
    print(f'\n {str1} and {str2} are permutations of each other')
    return sorted(str1)  == sorted(str2)


print(are_permutation("abc", "bca"))  # Output: True
print(are_permutation("abc", "def"))  # Output: False
```

4. Find the Second Largest Unique Number

Problem:
Given a list of integers, return the second largest unique number. If no second largest unique
number exists, return None.

Example:
Input: [4, 5, 6, 6, 7]
Output: 6

Explanation: The second largest unique number is 6.

```python
def second_largest_unique(nums):

  # remove duplicate use set
  unique_number = list(set(nums))

  if len(unique_number) < 2:
    print('No second largest unique number exists')
    return None

  unique_number.sort(reverse=True) # Sort in descending order
  second_largest_num = unique_number[1]
  print(f'The second largest unique number is {second_largest_num}')
  return second_largest_num

second_largest_unique([4, 5, 6, 6, 7])
second_largest_unique([20,20,20])
```

5. Count Distinct Pairs

Problem:

Given a list of integers and a target sum, count how many distinct pairs of integers in the list
sum up to the target.

Example:

Input: [1, 2, 3, 4, 3, 6], Target: 6

Output: 2

Explanation: The pairs (2, 4) and (1, 5) sum up to 6.

```python
def count_distinct_pairs(nums, target):
    seen = set()
    pairs = set()

    for num in nums:
        complement = target - num
        if complement in seen:

            pair = tuple(sorted((num, complement)))
            pairs.add(pair)
        seen.add(num)

    print(f"Number of distinct pairs that sum to {target}: {len(pairs)}")
    return len(pairs)


count_distinct_pairs([1, 2, 3, 4, 3, 6], 6)
```

6. Remove Duplicates from List

Problem:

Given a list of integers, remove all duplicates while maintaining the order of their first Occurrence.

Example:

Input: [1, 2, 2, 3, 4, 5, 1]
Output: [1, 2, 3, 4, 5]

Explanation: Duplicates are removed, but the order of elements is preserved.

```python
def remove_duplicates(nums):

  unique_check = set()
  result = []

  for num in nums:
    if num not in unique_check:
      unique_check.add(num)
      result.append(num)
  print(f"list of integers, remove all duplicates : {result}")
  return result


remove_duplicates([1, 2, 2, 3, 4, 5, 1])
```

7. Subsets of a List

Problem:

Given a list of integers, return all possible subsets (the power set) of the list.

Example:

Input: [1, 2, 3]

Output: [[], [1], [2], [3], [1, 2], [1, 3], [2, 3], [1, 2, 3]]

Explanation: The function returns all subsets of the list.

```python
def get_subsets(nums):
  result = [[]]

  for num in nums:
    new_subset = []
    for subset in result:
      new_subset.append(subset + [num])
    result.extend(new_subset)

  return result

get_subsets( [1, 2, 3])
```

8. Longest Increasing Subsequence

Problem:

Given a list of integers, find the length of the longest subsequence of strictly increasing Elements.

Example:

Input: [10, 9, 2, 5, 3, 7, 101, 18]

Output: 4

Explanation: The longest increasing subsequence is [2, 3, 7, 101].

```python
def longest_inreasing_subsquence(nums):

  if not nums:
    print(f'List is Empty')
    return 0

  lengths = [1] * len(nums)

  for i in range(len(nums)):
    for j in range(i):
      if nums[i] > nums[j]:
        lengths[i] = max(lengths[i], lengths[j] + 1)

  list_length = max(lengths)
  print(f"length of the longest increasing subsequence")
  return list_length

longest_inreasing_subsquence([10, 9, 2, 5, 3, 7, 101, 18])
```

9. Maximum Product of Two Numbers

Problem:

Given a list of integers, return the maximum product of any two numbers in the list.

Example:

Input: [1, 10, -5, 1, 100]

Output: 1000

Explanation: The largest product comes from 10 * 100 = 1000.

```python
def max_product(nums):
    if len(nums) < 2:
        return "Need at least two numbers for a product."

    nums.sort()

    # Product of two largest numbers OR two smallest (in case of negative * negative)
    product1 = nums[-1] * nums[-2]
    product2 = nums[0] * nums[1]

    max_product = max(product1, product2)

    return f"The maximum product of any two numbers is: {max_product}"


max_product([1, 10, -5, 1, 100])
```

10. Frequency of Characters in String

Problem:

Given a string, return the character that appears most frequently. If there is a tie, return the lexicographically smallest character.

Example:

Input: "aabbbc"

Output: 'b'

Explanation: The most frequent character is 'b'
.

```python
def most_frequent_char(s):
    # Count frequency of each character
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1

    # Find the character(s) with the highest frequency
    max_freq = max(freq.values())
    candidates = [char for char, count in freq.items() if count == max_freq]

    # Return the lexicographically smallest one among them
    result = min(candidates)
    return f"The most frequent character is: '{result}'"

# Test cases
print(most_frequent_char("aabbbc"))
print(most_frequent_char("aabbccddeeff"))
```

11. Print Prime Numbers Less Than N

Problem:

Write a function that prints all prime numbers less than a given number n.

Example:

Input: 10

Output: 2, 3, 5, 7

```python
def print_primes_less_than_n(n):
    print(f"Prime numbers less than {n} are:")
    for num in range(2, n):
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num, end=", ")
    print() # for a clean newline


print_primes_less_than_n(10)
```

13. Find the Student with Highest Score

Problem:

Given a dictionary of students' names and their scores, write a function that returns the name of
the student with the highest score.

Example:

Input: {"Alice": 85, "Bob": 90, "Charlie", 92}

Output: "Charlie"


```python
def highest_scoring_student(scores_dict):
    if not scores_dict:
        return "No student data available."

    highest_student = max(scores_dict, key=scores_dict.get)
    print(f"Student with the highest score is: {highest_student}")
    return highest_student


student_scores = {"Alice": 85, "Bob": 90, "Charlie": 92}
result = highest_scoring_student(student_scores)
```

15. Sum or Multiply Based on Keyword Argument

Problem:

Write a function that takes a variable-length list of numbers (*args) and an optional keyword
argument operation. If operation='multiply'
, return the product of all the numbers;
otherwise, return their sum.

Example:

Input: (1, 2, 3, 4), operation='multiply'

Output: 24

Explanation: The product of the numbers 1 * 2 * 3 * 4 is 24.

```python
def sum_or_multiply(*args, operation='sum'):
    if not args:
        return "No numbers provided."

    if operation == 'multiply':
        result = 1
        for num in args:
            result *= num
        return f"Product is: {result}"
    else:
        result = sum(args)
        return f"Sum is: {result}"

print(sum_or_multiply(1, 2, 3, 4, operation='multiply'))
print(sum_or_multiply(1, 2, 3, 4, operation='sum'))
```
