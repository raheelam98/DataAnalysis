1. . Largest Difference in List
Problem:

Given a list of integers, return the largest difference between any two elements, where the
larger element appears after the smaller one. If no such pair exists, return 0.

Example:
Input: [1, 2, 90, 10, 110]
Output: 109

Explanation: The largest difference is between 1 and 110.

get the diff between two number
list is [1, 2, 90, 10, 110]

```python
def largest_difference(lists):
    """
    Finds the largest difference between two elements in an array where the larger element
    appears after the smaller one.

    Args:
      arr: A list of integers.

    Returns:
      The largest difference found, or 0 if no such pair exists.
    """
    if not lists or len(lists) < 2:
        return 0

    Max_Diff = 0
    Min_Val = lists[0]

    for i in range(1, len(lists)):
        if lists[i] > Min_Val:
            Max_Diff = max(Max_Diff, lists[i])
print(largest_difference([1, 2, 90, 10, 110]))
#pls q1. Result
```
