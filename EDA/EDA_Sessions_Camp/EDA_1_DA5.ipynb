# Task
Analyze the data in "/content/sample_data/ai_adoption_dataset.csv" using pandas, including reading the data, performing basic analysis, grouping, and binning.

## Load the dataset

### Subtask:
Load the dataset from the provided CSV file into a pandas DataFrame.


**Reasoning**:
Import pandas and load the CSV file into a DataFrame named `df`.



```python
import pandas as pd

df = pd.read_csv("/content/sample_data/ai_adoption_dataset.csv")
```

## Perform basic data analysis

### Subtask:
Display the first few rows, check the data types of columns, and get a summary of the dataset's statistics.


**Reasoning**:
Display the first few rows, check the data types, and get a statistical summary of the DataFrame as requested by the instructions.



```python
display(df.head())
display(df.info())
display(df.describe())
```

## Group data

### Subtask:
Group the data based on relevant columns to analyze relationships and trends within different categories.


**Reasoning**:
Group the DataFrame as requested in the instructions and display the head of the resulting DataFrames.



```python
adoption_by_country = df.groupby('country')['adoption_rate'].mean().reset_index()
users_by_industry_ai = df.groupby(['industry', 'ai_tool'])['daily_active_users'].sum().reset_index()
adoption_by_year_age = df.groupby(['year', 'age_group'])['adoption_rate'].mean().reset_index()

display(adoption_by_country.head())
display(users_by_industry_ai.head())
display(adoption_by_year_age.head())
```

## Bin data

### Subtask:
Apply binning to numerical columns to categorize data into groups, which can be useful for frequency analysis or further grouping.


**Reasoning**:
Apply equal-width binning to the 'adoption_rate' and 'daily_active_users' columns and display the first few rows of the modified DataFrame.



```python
df['adoption_rate_binned'] = pd.cut(df['adoption_rate'], bins=5, include_lowest=True)
df['daily_active_users_binned'] = pd.cut(df['daily_active_users'], bins=10, include_lowest=True)
display(df.head())
```
